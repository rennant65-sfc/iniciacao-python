```python
import pandas as pd

# 1. Simulação de um relatório de vendas enviado pela empresa
dados_vendas = {
    'Data': ['2026-04-01', '2026-04-01', '2026-04-02', '2026-04-02', '2026-04-03'],
    'Produto': ['Notebook', 'Mouse Gamer', 'Notebook', 'Teclado Mecânico', 'Mouse Gamer'],
    'Quantidade': [2, 5, 1, 3, 4],
    'Valor_Unitario': [3500.00, 150.00, 3500.00, 250.00, 150.00]
}

# 2. Criando o DataFrame (a "tabela" inteligente do pandas)
df = pd.DataFrame(dados_vendas)

# 3. Criando uma nova coluna calculada: Total por Venda
df['Total_Venda'] = df['Quantidade'] * df['Valor_Unitario']

# 4. Processamento dos Indicadores Financeiros (KPIs)
faturamento_total = df['Total_Venda'].sum()
total_itens_vendidos = df['Quantidade'].sum()
ticket_medio = df['Total_Venda'].mean()
produto_mais_vendido = df.groupby('Produto')['Quantidade'].sum().idxmax()

# 5. Exibição do Relatório Financeiro Executivo
print("=" * 45)
print("       RELATÓRIO DE DESEMPENHO FINANCEIRO     ")
print("=" * 45)
print(f"Faturamento Total:       R$ {faturamento_total:,.2f}")
print(f"Total de Itens Vendidos: {total_itens_vendidos} unidades")
print(f"Ticket Médio por Venda:  R$ {ticket_medio:,.2f}")
print(f"Produto Mais Vendido:    {produto_mais_vendido}")
print("=" * 45)
print("\nDetalhamento das Vendas:")
print(df[['Data', 'Produto', 'Quantidade', 'Total_Venda']])
```
