```python
altura = float(input("Digite a sua altura em metros: "))
peso_ideal = (72.7 * altura) - 58
print("O seu peso ideal é:", peso_ideal)
```

```python
celsius = float(input("Digite a temperatura em Celsius: "))
fahrenheit = (9 / 5) * celsius + 32
print("A temperatura em Fahrenheit é:", fahrenheit)
```

```python
salario_base = 1200.00
total_vendido = float(input("Digite o total vendido no mês de abril: "))
comissao = total_vendido * 0.10
salario_bruto = salario_base + comissao

print("Comissão do mês: R$", comissao)
print("Salário Bruto total: R$", salario_bruto)
```

```python
num1 = float(input("Digite o primeiro número: "))
operacao = input("Digite a operação desejada (+, -, *, /): ")
num2 = float(input("Digite o segundo número: "))

if operacao == "+":
    resultado = num1 + num2
elif operacao == "-":
    resultado = num1 - num2
elif operacao == "*":
    resultado = num1 * num2
elif operacao == "/":
    resultado = num1 / num2
else:
    resultado = "Operação inválida"

print("O resultado da operação é:", resultado)
```
