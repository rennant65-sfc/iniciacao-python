```python
x = -35.59
y = 35e3
z = 5j

print(type(x))
print(type(y))
print(type(z))
```

```python
# Converter de inteiro para decimal (float)
x = float(1)

# Converter de decimal para inteiro (int)
y = int(2.8)

# Converter de inteiro para complexo (complex)
z = complex(1)

print(x)
print(y)
print(z)
print(type(x))
print(type(y))
print(type(z))
```

```python
x = 5
y = 3

# 5 é diferente de 3
print(x != y)

# 5 é maior que 3
print(x > y)

# 5 é maior ou igual a 3
print(x >= y)

# 5 não é menor nem igual a 3
print(not (x <= y))
```
